# Data exploration
Inspect the locally cached adjusted close panel and split boundaries. This notebook never downloads data implicitly.

```python
from pathlib import Path
import pandas as pd
root = Path.cwd()
prices = pd.concat([pd.read_csv(root / f'data/processed/{s}_prices.csv', index_col='date', parse_dates=['date']).assign(split=s) for s in ['train','validation','test']])
prices.groupby('split').agg(['min','max','count'])
```

```python
clean = prices.drop(columns='split')
clean.describe().T
```

```python
clean.plot(title='Adjusted prices (local cached data)', figsize=(12, 5));
```
