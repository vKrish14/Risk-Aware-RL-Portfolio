# Causal feature analysis
Features use only current and trailing prices. The fitted scaler is trained only on the training period.

```python
from pathlib import Path
import pandas as pd
from src.features.feature_engineer import engineer_features
root = Path.cwd()
prices = pd.concat([pd.read_csv(root / f'data/processed/{s}_prices.csv', index_col='date', parse_dates=['date']) for s in ['train','validation','test']]).sort_index()
features = engineer_features(prices)
features.describe().T
```

```python
features.filter(like='SPY').plot(subplots=True, figsize=(12, 10), title='SPY causal features');
```
